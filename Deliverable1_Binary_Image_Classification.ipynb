# Deliverable 1 – Binary Image Classification

## Objective
The objective of this deliverable is to design a neural network-based
binary classification system capable of distinguishing photographs
from non-photographic images in an unlabeled dataset.


```python
import os
import shutil

base_dir = "/kaggle/input/datasets/ranimbenamara/dataset-livrable-1"
binary_dir = "/kaggle/working/binary_dataset"

photo_dir = os.path.join(binary_dir, "photo")
non_photo_dir = os.path.join(binary_dir, "non_photo")

os.makedirs(photo_dir, exist_ok=True)
os.makedirs(non_photo_dir, exist_ok=True)

def copy_or_create_empty(src, dst, label):
    if not os.path.exists(src):
        print(f"Warning: {label} folder not found at {src}; creating empty destination folder.")
        os.makedirs(dst, exist_ok=True)
        return
    shutil.copytree(src, dst, dirs_exist_ok=True)

photo_source = os.path.join(base_dir, "Dataset Livrable 1 - Photo")
text_source = os.path.join(base_dir, "Dataset Livrable 1 - Text")
schematics_source = os.path.join(base_dir, "Dataset Livrable 1 - Schematics")

print("Photo exists:", os.path.exists(photo_source))
print("Text exists:", os.path.exists(text_source))
print("Schematics exists:", os.path.exists(schematics_source))

copy_or_create_empty(photo_source, photo_dir, "Photo")
copy_or_create_empty(text_source, non_photo_dir, "Text")
copy_or_create_empty(schematics_source, non_photo_dir, "Schematics")

print("Binary dataset successfully created.")
```

```python
import os
import shutil

base_dir = "/kaggle/input/datasets/ranimbenamara/dataset-livrable-1"
binary_dir = "/kaggle/working/binary_dataset"

photo_dir = os.path.join(binary_dir, "photo")
non_photo_dir = os.path.join(binary_dir, "non_photo")

os.makedirs(photo_dir, exist_ok=True)
os.makedirs(non_photo_dir, exist_ok=True)

def copy_images_flat(src, dst, label):
    if not os.path.exists(src):
        print(f"Warning: missing source folder {src}")
        return
    copied = 0
    for root, _, files in os.walk(src):
        for f in files:
            if f.lower().endswith(('.jpg', '.jpeg', '.png')):
                src_path = os.path.join(root, f)
                rel_path = os.path.relpath(src_path, src)
                safe_name = rel_path.replace(os.sep, "__")
                target_name = f"{label}__{safe_name}"
                shutil.copy2(src_path, os.path.join(dst, target_name))
                copied += 1
    print(f"{label} copied:", copied)

copy_images_flat(
    os.path.join(base_dir, "Dataset Livrable 1 - Photo"),
    photo_dir,
    "photo"
)

copy_images_flat(
    os.path.join(base_dir, "Dataset Livrable 1 - Text"),
    non_photo_dir,
    "text"
)

copy_images_flat(
    os.path.join(base_dir, "Dataset Livrable 1 - Schematics"),
    non_photo_dir,
    "schematics"
)

print("Photos:", len([f for f in os.listdir(photo_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]))
print("Non-photos:", len([f for f in os.listdir(non_photo_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]))
```

```python
import os

base_dir = "/kaggle/input/datasets/ranimbenamara/dataset-livrable-1"
photo_source = os.path.join(base_dir, "Dataset Livrable 1 - Photo")
text_source = os.path.join(base_dir, "Dataset Livrable 1 - Text")
schematics_source = os.path.join(base_dir, "Dataset Livrable 1 - Schematics")

for name, path in {
    "Photo": photo_source,
    "Text": text_source,
    "Schematics": schematics_source
}.items():
    if not os.path.exists(path):
        print(f"{name} images: 0 (missing folder)")
        continue
    count = sum(
        1
        for root, _, files in os.walk(path)
        for f in files
        if f.lower().endswith(('.jpg', '.jpeg', '.png'))
    )
    print(f"{name} images:", count)
```

## 3. Build, train, and save the classifier

The final pipeline expects a saved binary classifier artifact, so this section trains a compact CNN on the prepared photo versus non-photo folders and writes the model to disk for reuse.

```python
import os
import tensorflow as tf

MODEL_OUTPUT_DIR = "/kaggle/working"
MODEL_PATH = os.path.join(MODEL_OUTPUT_DIR, "deliverable1.keras")
os.makedirs(MODEL_OUTPUT_DIR, exist_ok=True)

train_ds = tf.keras.utils.image_dataset_from_directory(
    binary_dir,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(128, 128),
    batch_size=32,
    label_mode="binary",
    class_names=["non_photo", "photo"],
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    binary_dir,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(128, 128),
    batch_size=32,
    label_mode="binary",
    class_names=["non_photo", "photo"],
)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().shuffle(1000).prefetch(AUTOTUNE)
val_ds = val_ds.cache().prefetch(AUTOTUNE)

classifier = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3)),
    tf.keras.layers.Rescaling(1.0 / 255),
    tf.keras.layers.Conv2D(32, 3, activation="relu", padding="same"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu", padding="same"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, 3, activation="relu", padding="same"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid"),
])

classifier.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

classifier.summary()

history = classifier.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
)

classifier.save(MODEL_PATH)
print("Saved classifier model to:", MODEL_PATH)
```
